# Phase 5: RFM Value Segmentation Analysis
**Project:** Customer Churn Prediction & Behavioral Segmentation  
**Objective:** Apply Recency, Frequency, and Monetary value segmentation adapted for subscription business models to prioritize customer cohorts by lifetime financial contribution and churn vulnerability.


```
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

DATA_PATH = os.path.join('..', 'data', 'processed', 'telco_engineered.csv')
if not os.path.exists(DATA_PATH):
    DATA_PATH = os.path.join('data', 'processed', 'telco_engineered.csv')

df = pd.read_csv(DATA_PATH)
print(f"Loaded engineered data: {df.shape}")
```

## 1. Execute RFM Pipeline

```
import sys
sys.path.append('..')
from src.rfm_analysis import compute_rfm_metrics, assign_rfm_segments, generate_rfm_summary_table

df_rfm = compute_rfm_metrics(df)
df_rfm = assign_rfm_segments(df_rfm)

summary = generate_rfm_summary_table(df_rfm)
display(summary)
```

## 2. RFM Segment Revenue & Churn Analysis

```
fig, axes = plt.subplots(1, 2, figsize=(16, 5.5))

# Revenue by segment
rev = df_rfm.groupby('RFM_Segment')['TotalCharges'].sum() / 1e6
rev.sort_values(ascending=False).plot(kind='bar', ax=axes[0], color='#2ca02c', edgecolor='black')
axes[0].set_title('Total Segment Revenue ($ Millions)', fontweight='bold')
axes[0].set_ylabel('$ Millions')
axes[0].tick_params(axis='x', rotation=25)

# Churn rate by segment
churn = df_rfm.groupby('RFM_Segment')['Churn'].apply(lambda x: (x == 'Yes').mean() * 100)
churn.loc[rev.sort_values(ascending=False).index].plot(kind='bar', ax=axes[1], color='#d62728', edgecolor='black')
axes[1].set_title('Churn Rate by Segment (%)', fontweight='bold')
axes[1].set_ylabel('Churn %')
axes[1].tick_params(axis='x', rotation=25)

plt.tight_layout()
plt.show()
```
