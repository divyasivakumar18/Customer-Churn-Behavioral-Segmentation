```
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual aesthetic
sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.figsize'] = (10, 5)

# Load dataset
DATA_PATH = os.path.join('..', 'data', 'raw', 'Telco-Customer-Churn.csv')
if not os.path.exists(DATA_PATH):
    DATA_PATH = os.path.join('data', 'raw', 'Telco-Customer-Churn.csv')

df = pd.read_csv(DATA_PATH)
print(f'Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns')
df.head()
```

## 1. Dataset Overview & Data Types

```
# Information on datatypes and nulls
df.info()
```

## 2. Missing Values & Whitespace Anomaly Detection
Check for explicit NaN values as well as hidden whitespace strings (e.g. ' ' in TotalCharges).

```
# Explicit nulls
print('Explicit Null Values per column:')
print(df.isnull().sum()[df.isnull().sum() > 0])

# Whitespace strings
whitespace_cols = {}
for col in df.columns:
    if df[col].dtype == object:
        ws_count = (df[col].astype(str).str.strip() == '').sum()
        if ws_count > 0:
            whitespace_cols[col] = ws_count

print('\nHidden Whitespace Blanks:')
print(whitespace_cols)

# Inspect rows with whitespace TotalCharges
spaces_df = df[df['TotalCharges'].astype(str).str.strip() == '']
print(f'\nCount of rows with blank TotalCharges: {len(spaces_df)}')
display(spaces_df[['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges', 'Contract', 'Churn']])
```

## 3. Target Distribution: Churn Analysis

```
churn_counts = df['Churn'].value_counts()
churn_pct = df['Churn'].value_counts(normalize=True) * 100

print('Churn Counts:')
print(churn_counts)
print('\nChurn Percentages:')
print(churn_pct.round(2))

# Visualize Target Imbalance
fig, ax = plt.subplots(1, 2, figsize=(12, 4.5))

churn_counts.plot(kind='bar', color=['#2b5c8f', '#d95f02'], ax=ax[0])
ax[0].set_title('Customer Churn Count', fontsize=12, fontweight='bold')
ax[0].set_xlabel('Churn Status')
ax[0].set_ylabel('Number of Customers')
ax[0].grid(axis='y', linestyle='--', alpha=0.7)

ax[1].pie(churn_pct, labels=churn_pct.index, autopct='%1.1f%%', colors=['#2b5c8f', '#d95f02'], startangle=90, explode=[0, 0.08])
ax[1].set_title('Churn Rate Proportions', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
```
