# Phase 4: Feature Engineering Pipeline
**Project:** Customer Churn Prediction & Behavioral Segmentation  
**Objective:** Create high-signal domain features (tenure cohorts, service bundles, financial discrepancies, payment habits) without data leakage.


```
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

DATA_PATH = os.path.join('..', 'data', 'processed', 'telco_cleaned.csv')
if not os.path.exists(DATA_PATH):
    DATA_PATH = os.path.join('data', 'processed', 'telco_cleaned.csv')

df = pd.read_csv(DATA_PATH)
print(f"Loaded cleaned data: {df.shape}")
```

## 1. Feature Engineering Transformation

```
import sys
sys.path.append('..')
from src.feature_engineering import create_engineered_features

df_feat = create_engineered_features(df)
print(f"Engineered dataset shape: {df_feat.shape}")
df_feat[['tenure_group', 'total_services_count', 'security_support_count', 'is_autopay', 'cost_per_service', 'Churn']].head(10)
```

## 2. Behavioral Insights from Engineered Features

```
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Churn by Tenure Cohort
tg_churn = df_feat.groupby('tenure_group', observed=False)['Churn'].value_counts(normalize=True).unstack() * 100
tg_churn[['No', 'Yes']].plot(kind='bar', ax=axes[0], color=['#1f77b4', '#d62728'], edgecolor='black')
axes[0].set_title('Churn Rate by Tenure Cohort (%)', fontweight='bold')
axes[0].set_ylabel('Percentage (%)')

# Churn by Total Services Count
sc_churn = df_feat.groupby('total_services_count')['Churn'].value_counts(normalize=True).unstack() * 100
sc_churn[['No', 'Yes']].plot(kind='bar', ax=axes[1], color=['#1f77b4', '#d62728'], edgecolor='black')
axes[1].set_title('Churn Rate by Total Subscribed Services (%)', fontweight='bold')
axes[1].set_ylabel('Percentage (%)')

plt.tight_layout()
plt.show()
```
